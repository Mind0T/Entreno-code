````python
#imprimir ventas, total, y promedio de ventas

ventas=[1200,850,1500,920,1100]
total=0
promedio=0
print("Las siguientes fueron las ventas")
for i, venta in enumerate(ventas):
    print(f"{f'Venta {i} -':<30}{f'$ {venta:>5,}'}")
for x in ventas:
    total+=x
promedio=total/len(ventas)
print(f"{f'Total venta:':<30}$ {total:>5,}")
print(f"{'El promedio por venta fue:':<30}$ {int(promedio):>5,}")
````

## Sesión 3 — Estructuras de datos

Python tiene 4 estructuras de datos fundamentales. Cada una sirve para algo diferente:

| Estructura | Ordenada | Mutable | Duplicados | Cuándo usarla |
|---|---|---|---|---|
| **Lista** `[]` | Sí | Sí | Sí | Colección general, secuencias |
| **Tupla** `()` | Sí | No | Sí | Datos que no deben cambiar |
| **Set** `{}` | No | Sí | No | Valores únicos, pertenencia rápida |
| **Diccionario** `{k:v}` | Sí* | Sí | Llaves no | Pares llave-valor, búsqueda por nombre |

\* Los diccionarios mantienen el orden de inserción desde Python 3.7+
